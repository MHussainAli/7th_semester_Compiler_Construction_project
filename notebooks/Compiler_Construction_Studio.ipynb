```python
from src.lexer.lexer import Lexer
from src.semantic.semantic_analyzer import SemanticAnalyzer
from src.intermediate.three_address_code import TACGenerator
from src.optimization.optimizer import Optimizer
from src.codegen.target_code import TargetCodeGenerator
```

## 1. Source Program

```python
source='''int a = 10;\nint b = 20;\nint c = a + b * 2;\nprint(c);'''
print(source)
```

## 2. Lexical Analysis

```python
tokens=Lexer(source).tokenize()
for t in tokens: print(t)
```

## 3. Semantic Analysis / Symbol Table

```python
sem=SemanticAnalyzer()
print(sem.analyze(tokens))
```

## 4. Three Address Code

```python
tac=TACGenerator(tokens).generate()
for x in tac: print(x)
```

## 5. Optimization

```python
optimized=Optimizer().optimize(tac)
for x in optimized: print(x)
```

## 6. Target Code

```python
for x in TargetCodeGenerator().generate(optimized): print(x)
```
